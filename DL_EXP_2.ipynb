**Task A: VISUALIZATION OF ACTIVATION FUNCTIONS**

Implement and plot the following activation functions:

•	Sigmoid

•	Tanh

•	ReLU

**Compare their:**

•	Output range

•	Saturation regions

•	Gradient behavior

•	Computational efficiency

•	Typical applications

```python
import numpy as np
import matplotlib.pyplot as plt

print("Priya Dharshini R")
print("24BAD092")
print("Task A: Visualization of Activation Functions")

# Create input values
x = np.linspace(-10, 10, 400)

# Activation functions
sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)
relu = np.maximum(0, x)

# Derivatives
sigmoid_derivative = sigmoid * (1 - sigmoid)
tanh_derivative = 1 - np.tanh(x) ** 2
relu_derivative = np.where(x < 0, 0, 1)

# Plot activation functions
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.plot(x, sigmoid)
plt.title("Sigmoid Activation")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)

plt.subplot(1, 3, 2)
plt.plot(x, tanh)
plt.title("Tanh Activation")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)

plt.subplot(1, 3, 3)
plt.plot(x, relu)
plt.title("ReLU Activation")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)

plt.tight_layout()
plt.show()

# Plot gradient behavior
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.plot(x, sigmoid_derivative)
plt.title("Sigmoid Gradient")
plt.xlabel("Input")
plt.ylabel("Gradient")
plt.grid(True)

plt.subplot(1, 3, 2)
plt.plot(x, tanh_derivative)
plt.title("Tanh Gradient")
plt.xlabel("Input")
plt.ylabel("Gradient")
plt.grid(True)

plt.subplot(1, 3, 3)
plt.plot(x, relu_derivative)
plt.title("ReLU Gradient")
plt.xlabel("Input")
plt.ylabel("Gradient")
plt.grid(True)

plt.tight_layout()
plt.show()

# Comparison information
print("\nActivation Function Comparison")
print("\nSigmoid")
print("Output range: [0, 1]")
print("Saturation: Occurs at very negative and positive inputs")
print("Gradient behavior: Small gradients in saturation regions")
print("Computational efficiency: Lower than ReLU because of exponential calculation")
print("Typical applications: Binary classification output layer")

print("\nTanh")
print("Output range: [-1, 1]")
print("Saturation: Occurs at very negative and positive inputs")
print("Gradient behavior: Small gradients in saturation regions")
print("Computational efficiency: Lower than ReLU")
print("Typical applications: Hidden layers and recurrent neural networks")

print("\nReLU")
print("Output range: [0, infinity)")
print("Saturation: No saturation in the positive region")
print("Gradient behavior: Gradient is 1 for positive inputs and 0 for negative inputs")
print("Computational efficiency: High because it uses a simple maximum operation")
print("Typical applications: Hidden layers in CNNs and ANNs")

print("\nTask A completed successfully.")
```

**Task B: PERFORMANCE COMPARISON OF ACTIVATION FUNCTIONS**

Train identical ANN architectures using:

•	Sigmoid

•	Tanh

•	ReLU

**Record and compare:**

•	Training accuracy

•	Validation accuracy

•	Training loss

•	Validation loss

•	Number of epochs required for convergence
Analyze which activation function performs best and justify the observations.


```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Flatten
from tensorflow.keras.callbacks import EarlyStopping

print("Priya Dharshini R")
print("24BAD092")
print("Task B: Performance Comparison of Activation Functions")

# Load MNIST dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize pixel values
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Create validation set from training data
X_val = X_train[-10000:]
y_val = y_train[-10000:]

X_train_new = X_train[:-10000]
y_train_new = y_train[:-10000]

print("\nDataset Loaded Successfully")
print("Training samples:", len(X_train_new))
print("Validation samples:", len(X_val))
print("Test samples:", len(X_test))

# Build identical ANN architecture
def build_model(activation_function):
    model = Sequential([
        Input(shape=(28, 28)),
        Flatten(),
        Dense(128, activation=activation_function),
        Dense(64, activation=activation_function),
        Dense(10, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model

# Activation functions
activations = ["sigmoid", "tanh", "relu"]

history_dict = {}
results = {}

# Train models
for activation in activations:

    print("\nTraining with", activation.upper(), "activation")

    model = build_model(activation)

    early_stopping = EarlyStopping(
        monitor="val_loss",
        patience=2,
        restore_best_weights=True
    )

    history = model.fit(
        X_train_new,
        y_train_new,
        epochs=20,
        batch_size=64,
        validation_data=(X_val, y_val),
        callbacks=[early_stopping],
        verbose=1
    )

    history_dict[activation] = history.history

    epochs_required = len(history.history["loss"])

    final_train_acc = history.history["accuracy"][-1]
    final_val_acc = history.history["val_accuracy"][-1]
    final_train_loss = history.history["loss"][-1]
    final_val_loss = history.history["val_loss"][-1]

    results[activation] = {
        "train_accuracy": final_train_acc,
        "val_accuracy": final_val_acc,
        "train_loss": final_train_loss,
        "val_loss": final_val_loss,
        "epochs": epochs_required
    }

    print("\nResults for", activation.upper())
    print("Training Accuracy:", round(final_train_acc, 4))
    print("Validation Accuracy:", round(final_val_acc, 4))
    print("Training Loss:", round(final_train_loss, 4))
    print("Validation Loss:", round(final_val_loss, 4))
    print("Epochs Required for Convergence:", epochs_required)

# Display comparison table
print("\nFinal Comparison")
print("Activation | Train Acc | Val Acc | Train Loss | Val Loss | Epochs")

for activation in activations:
    r = results[activation]

    print(
        activation.upper(),
        "|",
        round(r["train_accuracy"], 4),
        "|",
        round(r["val_accuracy"], 4),
        "|",
        round(r["train_loss"], 4),
        "|",
        round(r["val_loss"], 4),
        "|",
        r["epochs"]
    )

# Training accuracy graph
plt.figure(figsize=(10, 6))

for activation in activations:
    plt.plot(
        history_dict[activation]["accuracy"],
        label=activation.upper()
    )

plt.title("Training Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Validation accuracy graph
plt.figure(figsize=(10, 6))

for activation in activations:
    plt.plot(
        history_dict[activation]["val_accuracy"],
        label=activation.upper()
    )

plt.title("Validation Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Training loss graph
plt.figure(figsize=(10, 6))

for activation in activations:
    plt.plot(
        history_dict[activation]["loss"],
        label=activation.upper()
    )

plt.title("Training Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training Loss")
plt.legend()
plt.grid(True)
plt.show()

# Validation loss graph
plt.figure(figsize=(10, 6))

for activation in activations:
    plt.plot(
        history_dict[activation]["val_loss"],
        label=activation.upper()
    )

plt.title("Validation Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid(True)
plt.show()

# Find best activation based on validation accuracy
best_activation = max(
    results,
    key=lambda x: results[x]["val_accuracy"]
)

print("\nBest Activation Function")
print(
    best_activation.upper(),
    "achieved the highest validation accuracy of",
    round(results[best_activation]["val_accuracy"], 4)
)

print("\nTask B completed successfully.")
```

**Task C: COMPARISON OF OPTIMIZATION ALGORITHMS**

Train the same ANN using:

•	Stochastic Gradient Descent (SGD)

•	Momentum

•	RMSProp

•	Adam

**Compare:**

•	Training loss

•	Validation loss

•	Convergence speed

•	Final training accuracy

•	Final validation accuracy

**Visualize:**

•	Loss vs Epoch

•	Accuracy vs Epoch

Discuss which optimizer converges faster and produces better generalization.



```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Flatten
from tensorflow.keras.optimizers import SGD, RMSprop, Adam
from tensorflow.keras.callbacks import EarlyStopping

print("Priya Dharshini R")
print("24BAD092")
print("Task C: Comparison of Optimization Algorithms")

# Load MNIST dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize data
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Create validation set
X_val = X_train[-10000:]
y_val = y_train[-10000:]

X_train_new = X_train[:-10000]
y_train_new = y_train[:-10000]

print("\nDataset Loaded Successfully")
print("Training samples:", len(X_train_new))
print("Validation samples:", len(X_val))
print("Test samples:", len(X_test))

# Build identical ANN architecture
def build_model(optimizer):

    model = Sequential([
        Input(shape=(28, 28)),
        Flatten(),
        Dense(128, activation="relu"),
        Dense(64, activation="relu"),
        Dense(10, activation="softmax")
    ])

    model.compile(
        optimizer=optimizer,
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model

# Define optimizers
optimizers = {
    "SGD": SGD(learning_rate=0.01),
    "Momentum": SGD(learning_rate=0.01, momentum=0.9),
    "RMSProp": RMSprop(learning_rate=0.001),
    "Adam": Adam(learning_rate=0.001)
}

history_dict = {}
results = {}

# Train using each optimizer
for optimizer_name, optimizer in optimizers.items():

    print("\nTraining with", optimizer_name)

    model = build_model(optimizer)

    early_stopping = EarlyStopping(
        monitor="val_loss",
        patience=2,
        restore_best_weights=True
    )

    history = model.fit(
        X_train_new,
        y_train_new,
        epochs=20,
        batch_size=64,
        validation_data=(X_val, y_val),
        callbacks=[early_stopping],
        verbose=1
    )

    history_dict[optimizer_name] = history.history

    epochs_required = len(history.history["loss"])

    final_train_loss = history.history["loss"][-1]
    final_val_loss = history.history["val_loss"][-1]
    final_train_acc = history.history["accuracy"][-1]
    final_val_acc = history.history["val_accuracy"][-1]

    results[optimizer_name] = {
        "train_loss": final_train_loss,
        "val_loss": final_val_loss,
        "train_accuracy": final_train_acc,
        "val_accuracy": final_val_acc,
        "epochs": epochs_required
    }

    print("\nResults for", optimizer_name)
    print("Training Loss:", round(final_train_loss, 4))
    print("Validation Loss:", round(final_val_loss, 4))
    print("Training Accuracy:", round(final_train_acc, 4))
    print("Validation Accuracy:", round(final_val_acc, 4))
    print("Convergence Epochs:", epochs_required)

# Display comparison
print("\nOptimizer Comparison")
print("Optimizer | Train Loss | Val Loss | Train Acc | Val Acc | Epochs")

for optimizer_name in optimizers:

    r = results[optimizer_name]

    print(
        optimizer_name,
        "|",
        round(r["train_loss"], 4),
        "|",
        round(r["val_loss"], 4),
        "|",
        round(r["train_accuracy"], 4),
        "|",
        round(r["val_accuracy"], 4),
        "|",
        r["epochs"]
    )

# Loss vs Epoch
plt.figure(figsize=(10, 6))

for optimizer_name in optimizers:
    plt.plot(
        history_dict[optimizer_name]["loss"],
        label=optimizer_name
    )

plt.title("Training Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training Loss")
plt.legend()
plt.grid(True)
plt.show()

# Validation loss vs Epoch
plt.figure(figsize=(10, 6))

for optimizer_name in optimizers:
    plt.plot(
        history_dict[optimizer_name]["val_loss"],
        label=optimizer_name
    )

plt.title("Validation Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid(True)
plt.show()

# Training accuracy vs Epoch
plt.figure(figsize=(10, 6))

for optimizer_name in optimizers:
    plt.plot(
        history_dict[optimizer_name]["accuracy"],
        label=optimizer_name
    )

plt.title("Training Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Validation accuracy vs Epoch
plt.figure(figsize=(10, 6))

for optimizer_name in optimizers:
    plt.plot(
        history_dict[optimizer_name]["val_accuracy"],
        label=optimizer_name
    )

plt.title("Validation Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Find best optimizer based on validation accuracy
best_optimizer = max(
    results,
    key=lambda x: results[x]["val_accuracy"]
)

# Find fastest convergence
fastest_optimizer = min(
    results,
    key=lambda x: results[x]["epochs"]
)

print("\nBest Optimizer")
print(
    best_optimizer,
    "achieved the highest validation accuracy of",
    round(results[best_optimizer]["val_accuracy"], 4)
)

print("\nFastest Convergence")
print(
    fastest_optimizer,
    "converged in",
    results[fastest_optimizer]["epochs"],
    "epochs"
)

print("\nTask C completed successfully.")
```
