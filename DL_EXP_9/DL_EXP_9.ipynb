```python
# PART A: Loading and Configuring a Pre-trained LLM

student_name = "Priya Dharshini R"
roll_number = "24BAD092"

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

tokenizer.pad_token = tokenizer.eos_token

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
model.eval()

print("Student Name:", student_name)
print("Roll Number:", roll_number)
print("Model Loaded Successfully!")
print("Device:", device)
```

```python
prompt = "What is Artificial Intelligence?"

inputs = tokenizer(prompt, return_tensors="pt").to(device)

with torch.no_grad():
    output = model.generate(
        **inputs,
        max_new_tokens=40,
        do_sample=True,
        temperature=0.7,
        pad_token_id=tokenizer.eos_token_id
    )

print(tokenizer.decode(output[0], skip_special_tokens=True))
```

**PART B: IMPLEMENTING CHATBOT FUNCTIONALITY**

**Perform the following tasks:**

•	Create a function to accept user queries and generate responses using the loaded LLM.

•	Construct suitable prompts and display the generated responses.

•	Maintain conversation history, if supported by the model.

•	Test the chatbot with different queries and observe response quality.



```python
# PART B: Implementing Chatbot Functionality

# Student Details
student_name = "Priya Dharshini R"
roll_number = "24BAD092"

# Function to generate chatbot response
def generate_response(user_query, conversation_history=None):

    prompt = ""

    # Prepare conversation history
    if conversation_history:
        for message in conversation_history[-6:]:

            if isinstance(message, dict):
                role = message.get("role")
                content = message.get("content", "")
                prompt += f"{role}: {content}\n"

            elif isinstance(message, (list, tuple)):
                prompt += f"User: {message[0]}\n"

                if len(message) > 1 and message[1]:
                    prompt += f"Assistant: {message[1]}\n"

    prompt += f"User: {user_query}\nAssistant:"

    # Tokenize input
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=900
    ).to(device)

    # Generate response
    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=80,
            do_sample=True,
            temperature=0.7,
            pad_token_id=tokenizer.eos_token_id
        )

    # Decode response
    response = tokenizer.decode(
        output[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return response.strip() or "Please try another question."


# Test chatbot
query = "What is Machine Learning?"

answer = generate_response(query, [])

# Display output
print("Student Name:", student_name)
print("Roll Number:", roll_number)
print("User Query:", query)
print("Chatbot Response:", answer)
```

**PART C: DEVELOPING AN INTERACTIVE INTERFACE**

**Perform the following tasks:**

•	Select Streamlit or Gradio to develop the chatbot interface.

•	Create a text input component and a submission mechanism.

•	Connect the interface to the LLM response-generation function.

•	Display responses in a chat-style layout.

•	Run and test the application with different user inputs.


```python
# PART C: Developing Interactive Chatbot Interface

# Import Gradio
import gradio as gr

# Chatbot function
def chatbot(message, history):
    try:
        response = generate_response(message, history)
        return response
    except Exception as e:
        return "Error: " + str(e)

# Create interactive chatbot interface
demo = gr.ChatInterface(
    fn=chatbot,
    title="AI Chatbot - Priya Dharshini R",
    description="Roll Number: 24BAD092 | Transformer-Based LLM Chatbot",
    examples=[
        "What is Artificial Intelligence?",
        "Explain Machine Learning.",
        "What is Deep Learning?",
        "What is Python?"
    ]
)

# Launch chatbot
demo.launch(share=True)
```

```python
%%writefile app.py

# IMPORTANT: Import spaces before torch
import spaces

import torch
import gradio as gr
from transformers import AutoTokenizer, AutoModelForCausalLM

# Student Details
student_name = "Priya Dharshini R"
roll_number = "24BAD092"

# Load GPT-2 model
model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

tokenizer.pad_token = tokenizer.eos_token

# Generate chatbot response
@spaces.GPU(duration=60)
def generate_response(message, history):

    prompt = ""

    for item in history[-6:]:
        if isinstance(item, dict):
            prompt += f"{item['role']}: {item['content']}\n"
        elif isinstance(item, (list, tuple)):
            prompt += f"User: {item[0]}\n"
            if len(item) > 1 and item[1]:
                prompt += f"Assistant: {item[1]}\n"

    prompt += f"User: {message}\nAssistant:"

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=900
    )

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=80,
            do_sample=True,
            temperature=0.7,
            pad_token_id=tokenizer.eos_token_id
        )

    response = tokenizer.decode(
        output[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return response.strip() or "Please try another question."


# Create Gradio interface
demo = gr.ChatInterface(
    fn=generate_response,
    title="AI Chatbot - Priya Dharshini R",
    description="Roll Number: 24BAD092 | Transformer-Based LLM Chatbot"
)

demo.launch()
```
