**Task1: Create a single layer perceptron class. Implement the AND, OR binary logic using the perceptron. improve the model accuracy.**

**Step 1:** Import the Required Library

**Step 2:** Create the Single Layer Perceptron Class

**Step 5:** Train the Perceptron for the AND Gate

**Step 6:** Display the AND Gate Predictions, Final Weights, and Accuracy

**Step 7:** Train the Perceptron for the OR Gate

**Step 8:** Display the OR Gate Predictions, Final Weights, and Accuracy


```python
import numpy as np

# Single Layer Perceptron Class
class Perceptron:
    def __init__(self, num_inputs, learning_rate=0.1):
        # Initialize weights (including bias)
        self.weights = np.random.rand(num_inputs + 1)
        self.learning_rate = learning_rate

    # Linear function
    def linear(self, inputs):
        return np.dot(inputs, self.weights[1:]) + self.weights[0]

    # Activation Function (Heaviside Step Function)
    def Heaviside_step_fn(self, z):
        if z >= 0:
            return 1
        else:
            return 0

    # Prediction
    def predict(self, inputs):
        z = self.linear(inputs)
        return self.Heaviside_step_fn(z)

    # Loss (Error)
    def loss(self, prediction, target):
        return target - prediction

    # Training
    def train(self, inputs, target):
        prediction = self.predict(inputs)
        error = self.loss(prediction, target)

        # Update weights
        self.weights[1:] += self.learning_rate * error * inputs

        # Update bias
        self.weights[0] += self.learning_rate * error

    # Fit Model
    def fit(self, X, y, epochs):
        for epoch in range(epochs):
            for inputs, target in zip(X, y):
                self.train(inputs, target)

    # Accuracy
    def accuracy(self, X, y):
        correct = 0
        for inputs, target in zip(X, y):
            if self.predict(inputs) == target:
                correct += 1
        return (correct / len(y)) * 100


# Student Details
print("Name     : Priya Dharshini R")
print("Roll No. : 24BAD092")
print("=" * 50)

# Dataset (Truth Table)
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# AND Gate
print("\nTraining Perceptron for AND Gate")

y_and = np.array([0, 0, 0, 1])

model_and = Perceptron(num_inputs=2, learning_rate=0.1)
model_and.fit(X, y_and, epochs=100)

print("\nFinal Weights (AND Gate):")
print(model_and.weights)

print("\nPredictions (AND Gate):")
for x in X:
    print(f"{x} -> {model_and.predict(x)}")

print("\nAccuracy (AND Gate): {:.2f}%".format(model_and.accuracy(X, y_and)))

# OR Gate
print("\n" + "=" * 50)
print("Training Perceptron for OR Gate")

y_or = np.array([0, 1, 1, 1])

model_or = Perceptron(num_inputs=2, learning_rate=0.1)
model_or.fit(X, y_or, epochs=100)

print("\nFinal Weights (OR Gate):")
print(model_or.weights)

print("\nPredictions (OR Gate):")
for x in X:
    print(f"{x} -> {model_or.predict(x)}")

print("\nAccuracy (OR Gate): {:.2f}%".format(model_or.accuracy(X, y_or)))
```

```python
import numpy as np
import pandas as pd

# Student Details
print("Name     : Priya Dharshini R")
print("Roll No. : 24BAD092")
print("=" * 50)


# Single Layer Perceptron Class

class Perceptron:

    def __init__(self, num_inputs, learning_rate=0.1):
        self.weights = np.random.rand(num_inputs + 1)
        self.learning_rate = learning_rate

    # Linear Function
    def linear(self, inputs):
        return np.dot(inputs, self.weights[1:]) + self.weights[0]

    # Activation Function
    def activation(self, z):
        if z >= 0:
            return 1
        else:
            return 0

    # Prediction
    def predict(self, inputs):
        z = self.linear(inputs)
        return self.activation(z)

    # Training
    def train(self, inputs, target):
        prediction = self.predict(inputs)
        error = target - prediction

        # Update weights
        self.weights[1:] += self.learning_rate * error * inputs

        # Update bias
        self.weights[0] += self.learning_rate * error

    # Fit Model
    def fit(self, X, y, epochs):
        for epoch in range(epochs):
            for inputs, target in zip(X, y):
                self.train(inputs, target)

    # Accuracy
    def accuracy(self, X, y):
        predictions = [self.predict(x) for x in X]
        return np.mean(np.array(predictions) == y) * 100

# Generate Random Dataset

np.random.seed(42)

# Generate 100 random samples with 2 binary features
X = np.random.randint(0, 2, (100, 2))

# Generate target using OR logic
y = np.where((X[:,0] == 1) | (X[:,1] == 1), 1, 0)

# Display Dataset
dataset = pd.DataFrame(X, columns=["Feature1", "Feature2"])
dataset["Target"] = y

print("\nRandom Dataset (First 10 Rows)")
print(dataset.head(10))

# Create Model

model = Perceptron(num_inputs=2, learning_rate=0.1)

# Train Model

print("\nTraining the Perceptron...")
model.fit(X, y, epochs=100)
print("Training Completed!")

# Predictions

print("\nPredictions (First 10 Samples)")
print("-" * 45)

for i in range(10):
    prediction = model.predict(X[i])
    print(f"Input: {X[i]}  Actual: {y[i]}  Predicted: {prediction}")

# Accuracy

accuracy = model.accuracy(X, y)

print("\nModel Accuracy: {:.2f}%".format(accuracy))

# Final Weights

print("\nFinal Weights:")
print(model.weights)
```
