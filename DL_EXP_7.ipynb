**PART A: DATASET PREPARATION AND TEXT PREPROCESSING**

**Perform the following tasks:**

•	Load the Brown Corpus available in NLTK and select a suitable text category.

•	Examine sample sentences from the selected dataset.

•	Convert the text into lowercase.

•	Perform word tokenization using NLTK.

•	Remove punctuation and stop words.

•	Perform lemmatization using WordNetLemmatizer.

•	Display the original and preprocessed text.


```python

!pip -q install nltk gensim

import nltk
import string
from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download("brown", quiet=True)
nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("omw-1.4", quiet=True)

# Load sentences from the Brown Corpus news category
original_sentences = brown.sents(categories="news")[:4000]

print("\nTotal original sentences:", len(original_sentences))
print("\nOriginal sentence:")
print(" ".join(original_sentences[0]))

# Preprocess the text
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()
clean_sentences = []

for sentence in original_sentences:
    text = " ".join(sentence).lower()
    words = word_tokenize(text, preserve_line=True)

    # Remove punctuation, numbers, and other non-alphabetic tokens
    words = [word for word in words if word.isalpha()]

    # Remove stop words
    words = [word for word in words if word not in stop_words]

    # Lemmatize words
    words = [lemmatizer.lemmatize(word) for word in words]

    if words:
        clean_sentences.append(words)

print("\nOriginal Text:")
print(" ".join(original_sentences[0]))

print("\nPreprocessed Text:")
print(" ".join(clean_sentences[0]))

print("\nNumber of processed sentences:", len(clean_sentences))
print("\nPriya Dharshini R | Roll No: 24BAD092")
```

**PART B: VOCABULARY GENERATION AND WORD FREQUENCY ANALYSIS**

**Perform the following tasks:**

•	Create a vocabulary from the preprocessed text.

•	Calculate word frequencies using NLTK's FreqDist.

•	Identify the ten most frequently occurring words.

•	Display the vocabulary size and frequency distribution.

•	Prepare the tokenized sentences for Word2Vec training.


```python

from nltk import FreqDist

# Combine all preprocessed words
all_words = [
    word for sentence in clean_sentences for word in sentence
]

# Calculate word frequencies
frequencies = FreqDist(all_words)

# Create vocabulary
vocabulary = set(all_words)

print("\nVocabulary Size:", len(vocabulary))

print("\nTop 10 Most Frequent Words:")
for word, count in frequencies.most_common(10):
    print(word, ":", count)

print("\nTotal Word Tokens:", len(all_words))
print("Unique Words:", len(vocabulary))

print("\nSample Tokenized Sentence:")
print(clean_sentences[0])

print("\n Priya Dharshini R | Roll No: 24BAD092")
```

**PART C: WORD EMBEDDING GENERATION USING WORD2VEC**

**Perform the following tasks:**

•	Train a Word2Vec model using Gensim on the preprocessed sentences.

•	Use the Skip-gram architecture with an embedding dimension of 100 and a context-window size of 5.

•	Generate embedding vectors for selected words.

•	Display the embedding vector and its dimensions.

•	Identify the five most similar words for a selected word using the trained model.


```python

from gensim.models import Word2Vec

# Train a Skip-gram Word2Vec model
model = Word2Vec(
    sentences=clean_sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=1,
    workers=2,
    epochs=20,
    seed=42
)

# Display an embedding vector
word = "government"

if word in model.wv:
    vector = model.wv[word]

    print("Priya Dharshini R | Roll No: 24BAD092")
    print("\nSelected Word:", word)
    print("\nEmbedding Vector:")
    print(vector)
    print("\nVector Dimensions:", len(vector))

    print("\nFive Most Similar Words:")
    for similar_word, score in model.wv.most_similar(word, topn=5):
        print(similar_word, ":", round(float(score), 4))
else:
    print("Selected word is not in the vocabulary.")
```

**PART D: WORD SIMILARITY ANALYSIS AND EXPERIMENT MANAGEMENT**

**Perform the following tasks:**

•	Select three meaningful word pairs from the trained vocabulary.

•	Calculate the cosine similarity between each pair using the Word2Vec model.

•	Compare the similarity scores and identify the most similar pair.

•	Analyze the relationship between contextual similarity and the generated word embeddings.

•	Document the observations and upload the completed notebook to GitHub.


```python

# Calculate cosine similarity for three meaningful word pairs
pairs = [
    ("government", "state"),
    ("president", "government"),
    ("school", "student")
]

results = []

print("\nCosine Similarity Results:")

for word1, word2 in pairs:
    if word1 in model.wv and word2 in model.wv:
        score = float(model.wv.similarity(word1, word2))
        results.append((word1, word2, score))
        print(f"{word1} - {word2}: {score:.4f}")
    else:
        print(f"Word pair not available: {word1}, {word2}")

if results:
    best = max(results, key=lambda item: item[2])

    print("\nMost Similar Word Pair:")
    print(best[0], "and", best[1])
    print("Cosine Similarity:", round(best[2], 4))

print("\nObservation:")
print("Word2Vec represents words as numerical vectors based on their")
print("context in the training sentences.")
print("Higher cosine similarity indicates that two word vectors")
print("point in more similar directions.")
print("The results depend on the corpus and training settings.")
print("\n Priya Dharshini R | Roll No: 24BAD092")
```
